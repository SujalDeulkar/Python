```python
#03 Write a Python program to count the number of vowels and consonants in a given string.
s = input("Enter string: ")
vowels = 0
consonant = 0

for ch in s:
    if ch in "aeiouAEIOU":
        vowels += 1
    elif ch.isalpha():
        consonant += 1

print("Vowels: ", vowels)
print("Consonant: ", consonant)
```

```python
#04 Write a Python program to count uppercase letters, lowercase letters, digits, and special characters in a
#string.
s = input("Enter string: ")

u = l = d = sp = 0

for x in s:
    if x.isupper():
        u += 1
    elif x.islower():
        l += 1
    elif x.isdigit():
        d += 1
    else:
        sp += 1

print("Uppercase = ", u)
print("Lowercase = ", l)
print("Digits = ", d)
print("Special = ", sp)
```

```python
#05 Write a Python program to check whether a given string is a palindrome.

s = input("Enter a string: ")

if s == s[::-1]:
    print("Palindrome")
else:
    print("Not a Palindrome")
```

```python
#Write a Python program to reverse a string without using slicing.

text = input("Enter a string: ")
def reverse_string(s):
    reversed_str = ""
    for char in s:
        reversed_str = char + reversed_str
    return reversed_str


print("Reversed string:", reverse_string(text))
```

```python
#Write a Python program to find the first occurrence and last occurrence of a character in a string.
string = "programming"
char = "r"

first = string.find(char)
last = string.rfind(char)

print("First occurrence:", first)
print("Last occurrence:", last)
```

```python
#Write a Python program to remove duplicate characters from a string while preserving the original order.
string = input("Enter a string: ")

result = ""

for char in string:
    if char not in result:
        result += char

print("String after removing duplicates:", result)
```

```python
#Write a Python program to find the longest word in a sentence.

sentence = input("Enter a sentence: ")

words = sentence.split()

if words:
    longest = max(words, key=len)
    print("Longest word:", longest)
    print("Length:", len(longest))
else:
    print("No words found.")
```

```python
#Write a Python program to capitalize the first letter of every word in a sentence without using title().
sentence = input("Enter a sentence: ")

words = sentence.split()
result = []

for word in words:
    result.append(word[0].upper() + word[1:])

print("Capitalized sentence:", " ".join(result))
```
